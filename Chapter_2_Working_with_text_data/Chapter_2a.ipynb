## TOKENIZER PIPELINE

**COMPLETE TRAINING DATASET → TOKENIZED TRAINING DATASET → VOCABULARY → TOKEN IDs**

1. **Complete training dataset:**
   - Split text into individual tokens.
   - All duplicates are removed.
   - These tokens are now unique tokens.

2. **Tokenized training dataset:**
   - Sort unique tokens in alphabetical order.

3. **Vocabulary:**
   - Assign token IDs to unique tokens based on position.

4. **Token IDs**
   - Output the token IDs.

Applying the basic tokenizer to "The Verdict": 
1. Text is stripped and whitespaces are removed.
2. Token duplicates are removed and remaining tokens are sorted alphabetically.
3. Create vocabulary and assign token IDs to unique tokens.

```python
preprocessed = re.split(r'([],.:;?_!"()\']|--|\s)', raw_text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
vocab = {token:integer for integer, token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break
```

**Implementing a complete tokenizer class.**
1. The initialization method loads the vocabulary and stores it as a class attribute. The method also creates an inverse vocabulary which maps token IDs to their text counterparts.
2. The encode method encodes text into token IDs.
3. The decode method decodes token IDs back into text.

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s, i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([],.:;?_!"()\']|--|\s)', text)
        preprocessed = [
            item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])

        text = re.sub(r'([],.:;?_!"()\']|--|\s)', r'\1', text)
        return text
```

Using the encode method to convert text into token IDs.

```python
tokenizer = SimpleTokenizerV1(vocab)
text = """"It's the last he painted, you know,"
        Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
```

Using the decode method to convert these IDs back into text.

```python
print(tokenizer.decode(ids))
```
